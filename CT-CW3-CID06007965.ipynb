```julia
using Combinatorics

# Calculate the decoding error probability of a binary Hamming code
function decoding_error_probability(n, p_e)
    # Calculate P_err,1: the probability of an error occurring (i.e. 2 or more errors)）
    p_err = 0.0
    
    # Calculate error probabilities from 2 to n and accumulate them
    for k in 2:n
        p_err += binomial(n, k) * p_e^k * (1 - p_e)^(n - k)
    end
    
    return p_err
end

n = 63 
p_e = A.Q9_1_c_data_p  
A.Q9_1_c_ans_p_err = decoding_error_probability(n, p_e)
```

```julia
function ham3(q, field_elements)
    num_cols = q + q^2 + 1 # The number of columns is q + q^2+1
    H = zeros(Int, 3, num_cols) 

    H[1, q +2:end] .= 1
    H[2, 2:q+1] .= 1
    #Second line: The first q column is 1, and subsequent ti is filled in blocks.
    for i in 1:q
        H[2, q + (i - 1) * q + 2 : q+1 + i * q] .= field_elements[i]
    end

    # The third line: the first q columns are t1 to tq, and the following are all possible permutations of (t1, t2, ..., tq)
    H[3, 2:q+1] .= field_elements
    for i in 1:q
        H[3, q + (i - 1) * q + 2 : q + i * q+1] .= field_elements
    end
    H[3,1] = 1;

    return H
end
q = 5
field_elements = [0,1, 2, 3, 4]
H3 = ham3(q, field_elements)
H = reverse(H3,dims=1)
A.Q9_2_a_ans_H = reverse(H3,dims=1)
#find the Matrix of G let G*H'= 0;
# specified column index
cols_to_extract = [1, 2, 7]
# The remaining columns form H2
H5 = H[:, setdiff(1:size(H, 2), cols_to_extract)]'
H5 = mod.(-1*H5,5)
rows, cols = 28, 31
# Construct G matrix
G2 = zeros(Int, rows, cols)
# Replace columns 1, 2, 7
indices_to_replace = [1,2,7]
for (i, col) in enumerate(indices_to_replace)
    G2[:, col] = H5[:, i]
end
# Fill remaining columns with unit vectors
current_index = 1
for col in 1:cols
    if !(col in indices_to_replace)
        G2[current_index, col] = 1
        current_index += 1
    end
end
A.Q9_2_b_ans_G = G2;
```

```julia
using Combinatorics

# Calculate the decoding error probability of a binary Hamming code
function decoding_error_probability(n, p_e)
    # Calculate P_err,1: the probability of an error occurring (i.e. 2 or more errors)）
    p_err = 0.0
    
    # Calculate error probabilities from 2 to n and accumulate them
    for k in 2:n
        p_err += binomial(n, k) * p_e^k * (1 - p_e)^(n - k)
    end
    
    return p_err
end

n = 31
p_e = A.Q9_2_c_data_p 
A.Q9_2_c_ans_p_err = decoding_error_probability(n, p_e)
```

```julia
q = 5  
n = 31  
d_min = 3  
p_e = A.Q9_2_e_data_p 
k = 28
max_erasures = floor(Int, (d_min - 1) / 2) 

# Calculate the probability of correct decoding when there are less than `max_erasures` erasures
prob_correct_decoding = 0.0
for i in 0:max_erasures
    prob_correct_decoding += binomial(n, i) * (p_e^i) * ((1 - p_e)^(n - i))
end

# Calculate error probability
P_err_2 = 1 - prob_correct_decoding
A.Q9_2_e_ans_p_err = P_err_2;
```
