```julia
function extended_gcd(a, b)
    if b == 0
        return (a, 1, 0)
    else
        (g, x, y) = extended_gcd(b, a % b)
        return (g, y, x - (a ÷ b) * y)
    end
end
function solve_congruence(a, b, n)  
    (g, x, _) = extended_gcd(a, n)

    if b % g != 0
        return Int64[]
    end
    x0 = mod(x * (b ÷ g), n)
    solutions = Int64[] 
    for i in 0:g-1
        push!(solutions, mod(x0 + i * (n ÷ g), n))
    end
    return solutions
end

answers = []
a_values1 = [A.Q4_1_a_data_a, A.Q4_1_b_data_a, A.Q4_1_c_data_a, A.Q4_1_d_data_a, A.Q4_1_e_data_a, A.Q4_1_f_data_a]
b_values1 = [A.Q4_1_a_data_b, A.Q4_1_b_data_b, A.Q4_1_c_data_b, A.Q4_1_d_data_b, A.Q4_1_e_data_b, A.Q4_1_f_data_b]
n_values1 = [A.Q4_1_a_data_n, A.Q4_1_b_data_n, A.Q4_1_c_data_n, A.Q4_1_d_data_n, A.Q4_1_e_data_n, A.Q4_1_f_data_n]
for i in 1:6
    solutions1 = solve_congruence(a_values1[i], b_values1[i], n_values1[i])
    push!(answers, solutions1)
end
A.Q4_1_a_ans_x = answers[1];
A.Q4_1_b_ans_x = answers[2];
A.Q4_1_c_ans_x = answers[3];
A.Q4_1_d_ans_x = answers[4];
A.Q4_1_e_ans_x = answers[5];
A.Q4_1_f_ans_x = answers[6];
```

#### Solution of question a:
<br>According to the Fibonacci numbers follow the recurrence relation:

$$ F_n = F_{n-1} + F_{n-2} $$

Using the Euclidean algorithm, we can know:

$$ \gcd(F_n,F_{n-1}) = \gcd(F_{n-1},F_n\mod F_{n-1}) $$

Substituting the Fibonacci recurrence relation \( $F_n = F_{n-1} + F_{n-2}$ \), we get:

$$ F_n \mod F_{n-1} = F_{n-2} $$

Thus, the gcd simplifies to:

$$ \gcd(F_n,F_{n-1}) = \gcd(F_{n-1},F_{n-2}) $$

We can continue this process recursively until we reach  :

$$ 
F_2 = 1 \quad \text{and} \quad F_1 = 1 
$$

Therefore, we conclude:

$$ \gcd(F_2, F_1) = 1 $$
Thus, for any \( n $\geq$ 1 \),


$$ \gcd(F_n, F_{n-1}) = 1 $$

```julia
# Definition of the Function
function modular_inverse(a::Int, n::Int)
    if gcd(a, n) != 1
        return nothing  # No inverse if a and n are not coprime
    end
    _, x, _ = gcdx(a, n)
    return x % n
end

function cofactor_matrix_mod(N::Matrix{Int}, n::Int)
    size_N = size(N, 1)
    cofactors = zeros(Int, size_N, size_N)
        for i in 1:size_N
        for j in 1:size_N
            # Get the minor matrix and compute its determinant modulo n
            minor = N[setdiff(1:size_N, i), setdiff(1:size_N, j)]
            det_minor = round(Int, mod(det(minor), n))  # Round and convert to Int
            cofactors[i, j] = (-1)^(i + j) * det_minor
        end
    end
        return cofactors
end

function inverse_modular2(N::Matrix{Int}, n::Int)

    det_N = round(Int, mod(det(N), n)) 

    if gcd(det_N, n) != 1
        return nothing 
    end

    cofactor_mat = cofactor_matrix_mod(N, n)
    adjugate = transpose(cofactor_mat)
    det_inv = modular_inverse(det_N, n)
    N_inv = (adjugate * det_inv) .% n
    N_inv = (N_inv .+ n) .% n
    
    return N_inv
end
A.Q4_5_a_ans = inverse_modular2(A.Q4_5_a_data_M, A.Q4_5_a_data_m);
A.Q4_5_b_ans = inverse_modular2(A.Q4_5_b_data_M, A.Q4_5_b_data_m);
A.Q4_5_c_ans = inverse_modular2(A.Q4_5_c_data_M, A.Q4_5_c_data_m);


```

```julia
# save your answers in Q5_3_a_ans_len_key, Q5_3_b_ans_key, Q5_3_b_ans_para
#QUESTION 1 figure out the answer of len_Key
using FreqTables
# Function to calculate shifted coincidence frequency
function shifted_coincidence_frequency(ciphertext::String, L::Int)
    N = length(ciphertext)
    scf_values = zeros(Float64, L)
    for l in 1:L
        matches = 0
        for j in 1:(N - l)
            if ciphertext[j] == ciphertext[l + j]
                matches += 1
            end
        end
        # Calculate scf(l)       
        scf_values[l] = matches / (N - l)
    end
    
    return scf_values
end
A.Q5_3_a_ans_len_key = argmax(shifted_coincidence_frequency(A.Q5_3_data_encrypted_para,8));
```

```julia
# save your answers in Q5_4_a_ans, Q5_4_b_ans. 
using LinearAlgebra
using Base
function encrypt(plain_text::String, A::Array{Int64, 2}, b::Array{Int64, 1})
    # Remove spaces and punctuation
    plain_text = filter(c -> isletter(c), plain_text) 
    plain_text = uppercase(plain_text) 
    # Mapping letters and numbers
    char_to_num(c) = Int(c) - Int('A')  # 'A' -> 0, 'B' -> 1, ..., 'Z' -> 25
    num_to_char(n) = Char(n + Int('A')) # 0 -> 'A', 1 -> 'B', ..., 25 -> 'Z'
    block_size = size(A, 2)
    padded_length = (length(plain_text) + block_size - 1) ÷ block_size * block_size
    padded_text = plain_text * repeat('A', padded_length - length(plain_text)) # Fill 'A'
    ciphertext = ""
    for i in 1:block_size:padded_length
        block = padded_text[i:min(i + block_size - 1, padded_length)]
        x = reshape([char_to_num(c) for c in block], length(block), 1)
        c = (A * x + b) .% 26
    # Convert ciphertext numbers into characters and concatenate them
        for num in c
            ciphertext *= num_to_char(Int(num)) 
        end
    end 
    return ciphertext
end
A.Q5_4_a_ans  = encrypt(A.Q5_4_plain_text, A.Q5_4_data_A, A.Q5_4_data_b);
```
