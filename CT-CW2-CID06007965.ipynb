#### 1. **For \( $m \leq 100$ \):**

Since 101 is prime, by Fermat's Little Theorem, we know:
$$2^{100} \equiv 1 \mod 101$$
This implies that \($ 2^m \mod 101$ \) is periodic with a period of 100.

For \( $m \leq 100 $\), we can directly compute \($ 2^m \mod 101 $\). If needed, we can improve efficiency by using the modular inverse, especially for large exponents.

For example, if \($ m = 99$ \), we have:

$$2^{99} \equiv 2^{100-1} = 2^{100} \cdot 2^{-1} \equiv 1 \cdot 2^{-1} \mod 101$$

Here, we need to calculate \( $2^{-1} \mod 101$ \), the modular inverse of 2 module of 101.

#### 1.1 **Modular Inverse Calculation**:

To compute \( $2^{-1} \mod 101 $\), we solve the congruence:

$$2x \equiv 1 \mod 101$$

Using the extended Euclidean algorithm, we find:

$$2^{-1} \equiv 51 \mod 101$$

Thus, \($2^{99} \equiv 51 \mod 101 $\).

For general \( $m \leq 100 $\), you can directly compute \( $2^m \mod 101 $\) or use the inverse to speed up the calculation when necessary.

#### 2. **For \( m > 100 \):**

When \( m > 100 \), because \($ 2^{100} \equiv 1 \mod 101 $\), we have periodicity. We can reduce \( m \) modulo 100, which gives:

$$m = 100k + r, \quad \text{where} \quad r = m \mod 100$$

Then:
$$2^m \mod 101 = 2^r \mod 101$$

Here, \( r \) is the remainder when \( m \) is divided by 100. Using this periodicity, we reduce the exponent to a smaller value.

For example:
- If \( m = A.Q6_2_data_m = 12345714607\), then \( $m \mod 100 = 7 $\), so:
  
  $$2^{12345714607} \mod 101 = 2^7 \mod 101$$

  We can directly compute \($ 2^7 \mod 101 = 27 \mod 101 $\).

### Summary:

1. **When \($ m \leq 100 $\):**
   - Compute \($ 2^m \mod 101 $\) directly.
   - If \( m \) is large, use the modular inverse to speed up the calculation, like \($ 2^{-1} \mod 101 $\).

2. **When \( m > 100 \):**
   - Reduce \( m \) modulo 100, then compute \($ 2^r \mod 101$ \), where \($ r = m \mod 100 $\).

```julia
using Primes

function prime_factors(n::BigInt)
    """Returns a set of the prime factors of n using the Primes library."""
    # Use the result returned by the factor function
    fact = factor(n) 
    factors = Set{BigInt}()
    # Extract prime factors
    for (base, exponent) in fact
        push!(factors, base) 
    end   
    return factors
end

function is_primitive_root(a::BigInt, p::BigInt)
    """Checks if a is a primitive root of prime p."""
    if !isprime(p) || gcd(a, p) ≠ 1
        return false  # p must be a prime number and a and p are relatively prime
    end
    # Calculate φ(p) = p - 1
    phi = p - 1
    #  Get the prime factors of φ(p)
    factors = prime_factors(phi)
    # Verify primitive root conditions
    for factor in factors
        if powermod(a, phi ÷ factor, p) == 1
            return false  # If the conditions are not met, it is not the original root
        end
    end
    return true 
end
A.Q6_3_a_ans_isprimitive = is_primitive_root(A.Q6_3_a_data_a, A.Q6_3_a_data_p)
```

```julia
function next_primitive_root(a::BigInt, p::BigInt)
    """Find the smallest primitive root of a prime number p greater than or equal to a"""
    x = a  # Start checking from a
    while true
        if is_primitive_root(x, p)
            return x 
        end
        x += 1 
    end
end
A.Q6_3_b_ans_x = next_primitive_root(A.Q6_3_b_data_a, A.Q6_3_b_data_p)
```

```julia
using Polynomials
using Combinatorics

function inv_mod(a, m)
    g, x, _ = gcdx(a, m)
    if g != 1
        throw(ArgumentError("Inverse does not exist for $a and $m"))
    end
    return mod(x, m) # Use mod to ensure the result is non-negative
end

# Lagrangian interpolation, calculate f(0)
function lagrange_interpolation(x_vals, y_vals, p, x)
    n = length(x_vals)
    result = BigInt(0)  # The initialization result is of type BigInt
    for i in 1:n
        li = BigInt(1)  # 计Calculate L_i(x)
        for j in 1:n
            if i != j
                # Calculate the numerator and denominator of L_i(x)
                numerator = (BigInt(x) - x_vals[j]) % p
                denominator = (x_vals[i] - x_vals[j]) % p
                li *= numerator * inv_mod(denominator, p)
                li = mod(li, p)
            end
        end
        # Multiply by the corresponding y value and add to the result
        li *= y_vals[i] % p
        li = mod(li, p)
        result += li
        result = mod(result, p)
    end
    return result  
end

# Use combinations to recover secrets
function find_secret_with_combinations(x_vals, y_vals, p)
    secrets = Dict{BigInt, Vector{Vector{Int}}}()  # Store f(0) and its corresponding combined index
    all_secrets = Vector{BigInt}()  # Used to store all f(0) values
    share_indices = collect(1:length(x_vals))  # Convert to array

    # Traverse all possible five-point combinations
    for combination in combinations(share_indices, 5)
        # Extract x and y values ​​from the given combination
        x_subset = [x_vals[i] for i in combination]  
        y_subset = [y_vals[i] for i in combination] 
        # Calculate f(0) using current combined interpolation
        secret = lagrange_interpolation(x_subset, y_subset, p, BigInt(0))

        # Add f(0) value to all_secrets list
        push!(all_secrets, secret)

        # Save combination into secrets dictionary
        if haskey(secrets, secret)
            push!(secrets[secret], combination)
        else
            secrets[secret] = [combination]
        end
    end

    return secrets, all_secrets
end

p = parse(BigInt, A.Q7_4_data_p)  
x_vals = parse.(BigInt, A.Q7_4_b_data_x_vector)  
y_vals = parse.(BigInt, A.Q7_4_b_data_y_vector)

# Get f(0) and combination index of all five-point combinations
secrets, all_secrets = find_secret_with_combinations(x_vals, y_vals, p)

# Print all calculated values ​​of f(0)
println("All calculated values ​​of f(0): ")
println(all_secrets)

# Find the secret f(0) that appears the most times and the corresponding combination
correct_secret, combinations_used = findmax(secrets)
println("The recovered shared secret is: $correct_secret")
println("The combination with the most occurrences is: $combinations_used")
A.Q7_4_b_ans_error_idx = 1
A.Q7_4_b_ans_secret = string(combinations_used)
```
