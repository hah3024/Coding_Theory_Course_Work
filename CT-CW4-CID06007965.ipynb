```julia
using LinearAlgebra

function generate_reed_solomon_matrices(n::BigInt, k::BigInt, p::BigInt, primitive_element::BigInt)
    # Generate the Vandermonde matrix for G
    x = [BigInt(primitive_element)^i % BigInt(p) for i in 0:n-1]  # Powers of primitive element mod p
    G = zeros(BigInt, k, n)  # Initialize generator matrix

    for i in 1:k
        for j in 1:n
            G[i, j] = x[j]^(i-1) % BigInt(p)  # (primitive_element^j)^(i-1) mod p
        end
    end
    # Construct parity-check matrix H
    # H is constructed from the null space of G in GF(p)
    if k < n
        H = zeros(BigInt, n - k, n)  # Initialize H
        for i in 1:(n-k)
            for j in 1:n
                H[i, j] = x[j]^(i) % BigInt(p)  # Transposed Vandermonde for null space
            end
        end
    else
        error("k must be less than n to construct a parity-check matrix H")
    end

    return mod.(G,37), mod.(H,37)
end
# Example usage
p = BigInt(A.Q10_1_data_p)
n = p - 1 
primitive_element = BigInt(A.Q10_1_data_a) # Primitive element
k1 = n - A.Q10_1_data_d_1 + 1
k2 = n - A.Q10_1_data_d_2 + 1

A.Q10_1_1_ans_G, A.Q10_1_1_ans_H = generate_reed_solomon_matrices(n, k1, p, primitive_element)
# println("Generator Matrix G:")
# println(G)
# println("\nParity Check Matrix H:")
# println(H)
```

```julia
using Nemo
# 定义有限域 GF(2^5) 和不可约多项式
R, x = polynomial_ring(GF(2), "x")  # 定义 GF(2) 上的多项式环
irreducible_poly = R(A.Q10_2_data_polynomial)  # 不可约多项式
F32, z = finite_field(irreducible_poly,"z")  # 定义 GF(32)
a = F32(R(A.Q10_2_data_primitive_el))
###
function generate_reed_solomon_matrices2(k, n, primitive_element)
    # Generate the Vandermonde matrix for G
    x = [(primitive_element)^i for i in 0:n-1]  # Powers of primitive element mod p
    MS_G = matrix_space(F32, k, n)
    MS_H = matrix_space(F32, n-k, n)
    G = MS_G()  # Initialize generator matrix
    for i in 1:k
        for j in 1:n
            G[i, j] = x[j]^(i-1)  # (primitive_element^j)^(i-1) mod p
        end
    end
    # Construct parity-check matrix H
    # H is constructed from the null space of G in GF(p)
    if k < n
        H = MS_H()  # Initialize H
        for i in 1:(n-k)
            for j in 1:n
                H[i, j] = x[j]^(i)  # Transposed Vandermonde for null space
            end
        end
    else
        error("k must be less than n to construct a parity-check matrix H")
    end

    return G,H
end
G,H = generate_reed_solomon_matrices2(21,31,a)
new_poly_matrix1 = Matrix{Vector{Int}}(undef, size(G, 1),size(G, 2))
new_poly_matrix2 = Matrix{Vector{Int}}(undef, size(H, 1),size(H, 2))
for i in 1:size(G, 1)
    for j in 1:size(G, 2)
        new_poly_matrix1[i, j] = (collect(coefficients(lift(polyZZ,G[i,j]))))
    end
end
for i in 1:size(H, 1)
    for j in 1:size(H, 2)
        new_poly_matrix2[i, j] = (collect(coefficients(lift(polyZZ,H[i,j]))))
    end
end
A.Q10_2_1_ans_G = new_poly_matrix1
A.Q10_2_1_ans_H = new_poly_matrix2
```
